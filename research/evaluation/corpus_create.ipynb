```python
import pandas as pd
chunks_dataset = pd.read_csv('dataset/chunks_dataset.csv')

new_chunks_dataset = chunks_dataset.copy()
new_chunks_dataset.drop(columns=['content_index', 'article_url', 'video_urls', 'image_urls'], inplace=True)
```

```python
new_chunks_dataset.info()
```

```python
new_chunks_dataset.drop_duplicates(inplace=True) 
new_chunks_dataset.info()
```

```python
new_chunks_dataset = new_chunks_dataset[new_chunks_dataset["category"] != "FAQ"]
new_chunks_dataset.info()
```

```python
new_chunks_dataset["name"] = new_chunks_dataset["category"] + "\n" + new_chunks_dataset["title"] + "\n" + "Trường Đại học Sư phạm Kỹ thuật Thành phố Hồ Chí Minh"
new_chunks_dataset.rename(columns={"content" : "full_text"}, inplace = True)
```

```python
corpus = pd.DataFrame(new_chunks_dataset[["name", "full_text"]])
corpus.head()
```

```python
corpus.to_csv("dataset/corpus.csv", encoding='utf-8-sig', index=False)
```
