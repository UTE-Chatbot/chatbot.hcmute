```python
import pandas as pd
import re 
```

```python
df = pd.read_csv("./raw_corpus.csv")
```

```python
cols_to_drop = ["status", "created_at", "updated_at", "file_path"]
df = df.drop(columns=cols_to_drop, errors="ignore")
```

```python
def text_preprocessing(text):
    if pd.isna(text):
        return text

    pattern = r"(---.*?---|<<<.*?>>>|###.*?###)"
    text = re.sub(pattern, "", text, flags=re.DOTALL)

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
for idx, row in df.iterrows():
    df.at[idx, "full_text"] = text_preprocessing(row["full_text"])
```
