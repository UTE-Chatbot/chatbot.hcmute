```python
import re
import os
from typing import Callable

def replace_markdown_images_paths(markdown_text: str, upload_func: Callable[[str], str]) -> str:
    """
    Find all image links in Markdown, upload the local images using upload_func,
    and replace the paths with the returned URLs.
    
    :param markdown_text: Original Markdown content
    :param upload_func: Function that takes a local file path and returns the uploaded URL
    :return: Updated Markdown with uploaded image URLs
    """
    # Regex to match Markdown images: ![alt](path)
    pattern = re.compile(r'(!\[[^\]]*\]\()([^\)]+)(\))')

    def replacer(match):
        alt_text = match.group(1)
        path = match.group(2)
        closing = match.group(3)

        # Only process local files
        if path and os.path.exists(path):
            uploaded_url = upload_func(path)
            return f"{alt_text}{uploaded_url}{closing}"
        else:
            return match.group(0)  # leave unchanged if not a local file

    return pattern.sub(replacer, markdown_text)

def upload_local_file(file_path: str) -> str:
    return "OK_CONDE"

markdown_content = result.markdown  # from datalab_sdk
updated_markdown = replace_markdown_images_paths(markdown_content, upload_local_file)

# updated_markdown is now Markdown (not HTML) with MinIO URLs
```

```python
import tempfile
import requests
from llama_cloud_services import LlamaParse

async def parse_document_from_url(url: str, api_key: str, language: str = "vi", num_workers: int = 4):
    response = requests.get(url)
    response.raise_for_status()
    
    with tempfile.NamedTemporaryFile(suffix=".pdf", delete=False) as tmp_file:
        tmp_file.write(response.content)
        tmp_path = tmp_file.name

    parser = LlamaParse(
        api_key=api_key,
        num_workers=num_workers,
        high_res_ocr=True,
        outlined_table_extraction=True,
        verbose=False,
        language=language
    )

    result = await parser.aparse(tmp_path)
    return result

api_key = "llx-3QGh6ep1It6XWRbHLIudDhABPU1Maj6poC5LLbI0rbY4ebIA"

result = await parse_document_from_url(source_url, api_key)
```
